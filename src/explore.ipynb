## 1. Dataset overview

```python
train_cats = glob(os.path.join(BASE_DIR, "data", "train", "cats", "*.jpg"))train_dogs = glob(os.path.join(BASE_DIR, "data", "train", "dogs", "*.jpg"))test_cats = glob(os.path.join(BASE_DIR, "data", "test", "cats", "*.jpg"))test_dogs = glob(os.path.join(BASE_DIR, "data", "test", "dogs", "*.jpg"))print("Train cats:", len(train_cats))print("Train dogs:", len(train_dogs))print("Test cats:", len(test_cats))print("Test dogs:", len(test_dogs))
```

## 2. Sample images of cats

```python
def show_images(file_list, title, n=9):    fig, axes = plt.subplots(3, 3, figsize=(9, 9))    axes = axes.ravel()    for i in range(n):        img = mpimg.imread(file_list[i])        axes[i].imshow(img)        axes[i].axis("off")        axes[i].set_title(f"{img.shape[1]}x{img.shape[0]}")    plt.suptitle(title, fontsize=14)    plt.tight_layout()    plt.show()show_images(train_cats[:9], "Sample cat images - varying sizes")
```

## 3. Sample images of dogs

## 5. Data Generators with KerasWe use ImageDataGenerator with flow_from_directory to load images in batches.

```python
import tensorflow as tffrom tensorflow.keras.preprocessing.image import ImageDataGeneratorIMG_H, IMG_W = 224, 224; BS = 32train_datagen = ImageDataGenerator(rescale=1./255, shear_range=0.2, zoom_range=0.2, horizontal_flip=True)test_datagen = ImageDataGenerator(rescale=1./255)trdata = train_datagen.flow_from_directory(os.path.join(BASE_DIR, "data", "train"), target_size=(IMG_H, IMG_W), batch_size=BS, class_mode="binary", shuffle=True)tsdata = test_datagen.flow_from_directory(os.path.join(BASE_DIR, "data", "test"), target_size=(IMG_H, IMG_W), batch_size=BS, class_mode="binary", shuffle=False)print("Class mapping:", trdata.class_indices)
```
